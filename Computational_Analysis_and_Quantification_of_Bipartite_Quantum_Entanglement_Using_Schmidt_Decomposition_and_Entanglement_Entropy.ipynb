# **Entanglement of Quantum States**
Quantum entanglement is a phenomenon in which two or more quantum particles become correlated such that their states cannot be described independently, even when separated by large distances. It is a fundamental resource in quantum computing, communication and sensing.

**Measurement:**
Entanglement is quantified using mathematical measures:
- *Schmidt Rank:* Identifies whether a pure bipartite state is entangled.
- *Von Neumann Entropy:* Measures the degree of entanglement in pure bipartite states.
- *Concurrence:* Quantifies entanglement between two qubits.
- *Negativity:* Measures entanglement, including in mixed states.

**Industrial Applications:**
- *Quantum Computing:* Quantum algorithms and error correction.
- *Quantum Communication:* Quantum teleportation and secure key distribution.
- *Quantum Sensing:* High-precision measurements and atomic clocks.
- *Quantum Networks:* Connecting distant quantum processors through entanglement distribution.

**Physical Representation of Entanglement:**

Entanglement can be physically represented using two interacting or jointly prepared quantum systems, such as photons.

Consider the Bell state:
$$
|\Phi^+\rangle=\frac{|00\rangle+|11\rangle}{\sqrt{2}}
$$

Its physical interpretation is:
- If the first qubit is measured as 0, the second qubit is also found in 0.
- If the first qubit is measured as 1, the second qubit is also found in 1.
- Each individual outcome has a probability of 50%.

SVD stands for **Singular Value Decomposition**. In NumPy's linear algebra module (numpy.linalg), breaks down a 2D matrix (A) into three simpler component matrices:


$$A = U\Sigma V^\dagger$$
U: Left singular vectors (AA†).

Σ: Singular values (Diagonal matrix of eigenvalues of AA†).

V†: Conjugate transpose of right singular vectors (Eigenvector).

**Schmidt Rank** is the number of nonzero Schmidt coefficients $(\lambda _{i})$ in the Schmidt decomposition of a state in a tensor product space $\mathcal{H}_A \otimes \mathcal{H}_B$. Equivalently, it is the matrix rank of the coefficient matrix when the state is written out in chosen orthonormal bases.

## Mathematical Relationships

**SVD:**

$$
C=U\Sigma V^\dagger
$$

**Eigenvalue and Schmidt coefficient relationship:**

$$
CC^\dagger=U\Sigma^2U^\dagger
$$

$$
s_i=\sqrt{\mu_i}
$$



**Schmidt Rank:**

$$
\text{Schmidt Rank}=
\text{Number of nonzero Schmidt coefficients}
$$

For a two-qubit system:

Schmidt Rank = 1: Product (separable) state.

Schmidt Rank = 2: Entangled state.

A higher Schmidt rank indicates that the state cannot be represented as a product of individual subsystem states.

The numerical threshold $10^{-10}$ is used in the code to ignore extremely small numerical values caused by floating-point precision.

**Reduced Density Matrix**

The reduced density matrix describes the quantum state of one subsystem when the other subsystem is not considered.

For a pure bipartite quantum state, the reduced density matrix of subsystem A is calculated using:

$$
\rho_A=AA^\dagger
$$

Where:

$A$ is the coefficient matrix.

$A^\dagger$ is its conjugate transpose.

$\rho_A$ is the reduced density matrix of subsystem A.

The reduced density matrix provides information about the quantum state and entanglement of the subsystem.

**Eigenvalues of the Reduced Density Matrix**

The eigenvalues of the reduced density matrix are related to the Schmidt coefficients.

$$
\rho_A=AA^\dagger
$$

The eigenvalues are obtained by solving:

$$
\det(\rho_A-\lambda I)=0
$$

The relationship between eigenvalues and Schmidt coefficients is:

$$
\lambda_i=s_i^2
$$

Therefore:

$$
s_i=\sqrt{\lambda_i}
$$

For a normalized quantum state, the sum of eigenvalues must equal one.

$$
\sum_i\lambda_i=1
$$

**Trace of the Reduced Density Matrix**

The trace is the sum of the diagonal elements of a matrix.

For a valid normalized quantum state:

$$
\operatorname{Tr}(\rho_A)=1
$$

The trace condition verifies that the total probability of the quantum state is normalized.

**Entanglement Entropy (Von Neumann Entropy)**

Entanglement entropy measures the *degree of entanglement* between two subsystems of a pure bipartite quantum state.

It is calculated using:

$$
S=-\operatorname{Tr}(\rho_A\log_2\rho_A)
$$

Using the Schmidt coefficients, the same expression becomes:

$$
S=-\sum_i s_i^2\log_2(s_i^2)
$$

Interpretation:

$S=0$: No entanglement (Product State).

$S=1$: Maximum entanglement for a two-qubit pure state.

$0<S<1$: Partial entanglement.

The logarithm is taken to base 2, so the entropy is expressed in bits.

**Overall Mathematical Interpretation**

The analysis combines SVD, reduced density matrices, eigenvalue calculations, Schmidt rank and entanglement entropy to characterize different quantum states.

Conclusion: By combining these mathematical methods, the quantum states can be classified as product, partially entangled or maximally entangled states. This provides a mathematical basis for comparing entanglement properties with quantum circuit simulations.

```python
states = {
    "Bell State:": np.array([1, 0, 0, 1])/np.sqrt(2),
    "Partial Entanglement:": np.array([2, 0, 0, 1])/np.sqrt(5),
    "Product State 1:": np.array([1, 0, 0, 0]),
    "Product State 2:": np.array([1, 0, 1, 0])/np.sqrt(2)
}
entropies = []
for name, psi in states.items():

    print(name)

    # Construct coefficient matrix
    A = psi.reshape(2, 2)

    #SVD
    U, S, Vh = np.linalg.svd(A)

    #Diagonal Matrix S
    rank = np.sum(S > 1e-10)

    rho_A = A @ A.conj().T

    print("Reduced Density Matrix:")
    print(rho_A)

    # Calculate eigenvalues
    eigenvalues = np.linalg.eigvalsh(rho_A)

    print("\nEigenvalues:")
    print(eigenvalues)

    # Check trace
    print("\nTrace:", np.round(np.trace(rho_A), 10))

    entropy = -np.sum(
        S[S > 1e-10]**2 *
        np.log2(S[S > 1e-10]**2)
    )
    entropies.append(max(0.0, entropy))

    print("\nSchmidt coefficients:", S)
    print("Schmidt rank:", rank)
    print("Entanglement entropy:", max(0.0, entropy))
    print("-------------------------------------------------------------------")
```

```python
# Plot graph
plt.figure(figsize=(8, 5))

plt.bar(states.keys(), entropies)

plt.ylabel("Entanglement Entropy (bits)")
plt.xlabel("Quantum States")
plt.title("Visual Comparison of Quantum Entanglement")

plt.ylim(0, 1.1)
plt.grid(axis='y', alpha=0.5)

plt.show()
```

# Entanglement Variation with Theta

## 1. Objective

This simulation investigates how entanglement between two qubits changes as the parameter $\theta$ varies from $0$ to $\pi/4$.

A two-qubit quantum state is defined as:

$$
|\psi(\theta)\rangle=
\cos(\theta)|00\rangle+
\sin(\theta)|11\rangle
$$

The parameter $\theta$ controls the probability amplitudes of the two computational basis states and consequently determines the degree of entanglement.

## 2. Mathematical Representation

The coefficient matrix of the quantum state is:

$$
A(\theta)=
\begin{bmatrix}
\cos(\theta)&0\\
0&\sin(\theta)
\end{bmatrix}
$$

Using Singular Value Decomposition (SVD), the Schmidt coefficients are:

$$
s_1=\cos(\theta),\quad
s_2=\sin(\theta)
$$

The corresponding eigenvalues of the reduced density matrix are:

$$
\lambda_1=\cos^2(\theta)
$$

$$
\lambda_2=\sin^2(\theta)
$$

## 3. Entanglement Entropy

The entanglement entropy is calculated using the von Neumann entropy formula:

$$
S=-\sum_i\lambda_i\log_2(\lambda_i)
$$

Substituting the eigenvalues:

$$
S(\theta)=
-\cos^2(\theta)\log_2(\cos^2(\theta))
-\sin^2(\theta)\log_2(\sin^2(\theta))
$$

This equation quantifies the entanglement between the two qubits.

## 4. Interpretation of the Graph

The graph represents the variation of entanglement entropy as $\theta$ increases from $0$ to $\pi/4$.

- **At $\theta=0$:**

$$
|\psi\rangle=|00\rangle,\quad S=0
$$

The system is in a product state with no entanglement.

- **For $0<\theta<\pi/4$:**

Both basis states contribute to the quantum superposition. The state becomes partially entangled, and entanglement entropy increases.

- **At $\theta=\pi/4$:**

$$
|\psi\rangle=
\frac{1}{\sqrt{2}}(|00\rangle+|11\rangle)
$$

$$
S=1\text{ bit}
$$

The system reaches a maximally entangled Bell state.

## 5. Inferences

1. Entanglement entropy increases monotonically as $\theta$ varies from $0$ to $\pi/4$.
2. At $\theta=0$, the system has zero entanglement.
3. At $\theta=\pi/4$, both Schmidt coefficients become equal, resulting in maximum entanglement.
4. The parameter $\theta$ controls the distribution of Schmidt coefficients and therefore determines the degree of entanglement.
5. The maximum entanglement entropy for a two-qubit pure state is one bit.

**Conclusion:** The simulation demonstrates that entanglement is not simply a binary property. It can vary continuously depending on the quantum state amplitudes. By adjusting $\theta$, the system transitions from a separable state to a maximally entangled state.

```python
theta_values = np.linspace(0, np.pi/4, 100)
entropy_values = []

for theta in theta_values:

    psi = np.array([
        np.cos(theta), 0, 0, np.sin(theta)
    ])

    A = psi.reshape(2, 2)

    S = np.linalg.svd(A, compute_uv=False)

    p = S**2
    p = p[p > 1e-10]

    entropy = -np.sum(p * np.log2(p))

    entropy_values.append(entropy)

plt.figure(figsize=(8, 5))

plt.plot(theta_values, entropy_values, linewidth=2)

plt.xlabel("Theta (radians)")
plt.ylabel("Entanglement Entropy (bits)")
plt.title("Entanglement Variation with Theta")

plt.grid(True)
plt.show()
```

## Entanglement Entropy of Three Quantum Circuits

The von Neumann entanglement entropy of a pure bipartite quantum state is calculated using the reduced density matrix:

$$
S(\rho_A)=-\mathrm{Tr}(\rho_A\log_2\rho_A)
$$

Equivalently, using the Schmidt coefficients:

$$
\boxed{S=-\sum_i\lambda_i^2\log_2(\lambda_i^2)}
$$

### 1. Product State

The circuit prepares:

$$
|\psi_1\rangle=|00\rangle
$$

Its Schmidt coefficients are:

$$
\lambda_1=1,\qquad \lambda_2=0
$$

Therefore:

$$
S_1=-1\log_2(1)=0
$$

$$
\boxed{S_1=0\text{ bits}}
$$

### 2. Partially Entangled State

The circuit prepares:

$$
|\psi_2\rangle=
\cos\theta|00\rangle+\sin\theta|11\rangle
$$

For:

$$
\theta=\frac{\pi}{8}
$$

The Schmidt coefficients are:

$$
\lambda_1=\cos\frac{\pi}{8}\approx0.9239
$$

$$
\lambda_2=\sin\frac{\pi}{8}\approx0.3827
$$

The entanglement entropy is:

$$
S_2=-\cos^2\theta\log_2(\cos^2\theta)
-\sin^2\theta\log_2(\sin^2\theta)
$$

Substituting $\theta=\pi/8$:

$$
\boxed{S_2\approx0.6009\text{ bits}}
$$

### 3. Maximally Entangled Bell State

The circuit prepares:

$$
|\Phi^+\rangle=
\frac{|00\rangle+|11\rangle}{\sqrt2}
$$

The Schmidt coefficients are:

$$
\lambda_1=\lambda_2=\frac{1}{\sqrt2}
$$

Therefore:

$$
S_3=-\frac12\log_2\frac12
-\frac12\log_2\frac12
$$

$$
\boxed{S_3=1\text{ bit}}
$$

### 4. Comparative Results

| Quantum State | Schmidt Rank | Entanglement Entropy |
|---|---:|---:|
| Product State | 1 | 0 bits |
| Partially Entangled State | 2 | 0.6009 bits |
| Bell State | 2 | 1 bit |

### Conclusion

The results demonstrate that entanglement entropy increases from zero for a product state to one bit for a maximally entangled Bell state.

Although the partially entangled state and Bell state both have Schmidt rank 2, their entanglement entropies are different. This demonstrates that Schmidt rank identifies the presence of entanglement, while von Neumann entropy quantifies its degree.

```python
def calculate_entropy(circuit):

    # Obtain quantum state from circuit
    state = Statevector.from_instruction(circuit)

    # Extract statevector
    psi = state.data

    # Convert into coefficient matrix
    A = psi.reshape(2, 2)

    # Calculate Schmidt coefficients
    S = np.linalg.svd(A, compute_uv=False)

    # Calculate Schmidt rank
    rank = np.sum(S > 1e-10)

    # Calculate entropy
    p = S**2
    p = p[p > 1e-10]

    entropy = -np.sum(p * np.log2(p))

    return psi, S, rank, entropy


results = {}

for name, circuit in circuits.items():

    psi, S, rank, entropy = calculate_entropy(circuit)

    results[name] = entropy

    print(name)
    print("Statevector:", np.round(psi, 4))
    print("Schmidt coefficients:", np.round(S, 4))
    print("Schmidt rank:", rank)
    print("Entanglement entropy:", round(entropy, 4),"\n")
```

```python
import matplotlib.pyplot as plt

# Entropy values calculated from Qiskit circuits
labels = list(results.keys())
values = list(results.values())

plt.figure(figsize=(9, 5))

bars = plt.bar(
    labels,
    values,
    width=0.5,
    edgecolor="black"
)

# Display entropy values above bars
for bar, value in zip(bars, values):
    plt.text(
        bar.get_x() + bar.get_width()/2,
        bar.get_height() + 0.025,
        f"{value:.4f}",
        ha="center",
        fontsize=11
    )

plt.xlabel("Quantum States", fontsize=12)
plt.ylabel("Entanglement Entropy (bits)", fontsize=12)

plt.title("Comparison of Entanglement in Qiskit Circuits")

plt.ylim(0, 1.15)
plt.yticks(np.arange(0, 1.1, 0.2))
plt.grid(axis="y", linestyle="--", alpha=0.3)

plt.tight_layout()
plt.show()
```
